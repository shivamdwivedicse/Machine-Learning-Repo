```python
from sklearn.model_selection import train_test_split 
```

```python
x = df.drop('species' , axis =1)
y= df['species']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn  = KNeighborsClassifier(n_neighbors = 5)
```

```python
model_knn.fit(x_train , y_train)
```

```python
model_knn.score(x_test , y_test)
```

```python
#now let me change my n_neighbors from 5 to 9 
model_knn  = KNeighborsClassifier(n_neighbors = 9) 
model_knn.fit(x_train , y_train) 
model_knn.score(x_test , y_test)
```

```python
#now let me change my n_neighbors from 5 to 13
model_knn  = KNeighborsClassifier(n_neighbors = 13) 
model_knn.fit(x_train , y_train) 
model_knn.score(x_test , y_test)
```

```python
from sklearn.svm import SVC
```

```python
model_svm = SVC(gamma='auto' , C=30 , kernel='rbf')  #c can be say as punishment 
model_svm.fit(x_train , y_train) 
model_svm.score(x_test , y_test)
```

```python
# let use a linear kernel 
model_svm = SVC(gamma='auto' , C=30 , kernel='linear')  #c can be say as punishment 
model_svm.fit(x_train , y_train) 
model_svm.score(x_test , y_test)
```

```python
# let use C=10 
model_svm = SVC(gamma='auto' , C=10 , kernel='linear')  #c can be say as punishment 
model_svm.fit(x_train , y_train) 
model_svm.score(x_test , y_test)
```

```python
#Now lets use the Grid Search CV
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
#after importing lets use hyper parameter tuning and creat grids 
#So to perform GridsearchCV we need to select model let us select model_svm 
classifire = GridSearchCV((model_svm),{
    "C" :[1,10,20,30] , 
    "kernel" :['rbf' , 'linear'] , 
}, cv=5,return_train_score = False)
```

```python
classifire.fit(x,y)
```

```python
classifire.cv_results_
```

```python
results = pd.DataFrame(classifire.cv_results_)
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

```python
#lets do same thing for KNN
classifire_knn = GridSearchCV(
    model_knn,
    {
        "n_neighbors": [3, 5, 7, 9, 11],
        "weights": ["uniform", "distance"],
        "metric": ["euclidean", "manhattan", "minkowski"]
    },
    cv=5,
    return_train_score=False
)
```

```python
classifire_knn.fit(x,y)
```

```python
results_knn = pd.DataFrame(classifire_knn.cv_results_)
results_knn
```

```python
results_knn[['param_n_neighbors', 'param_weights', 'param_metric', 'mean_test_score']]
```

**USE RANDOM SEARCH CV**

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifire_rdcv = RandomizedSearchCV((model_svm),{
    "C" :[1,10,20,30] , 
    "kernel" :['rbf' , 'linear'] , 
}, n_iter = 4 , cv=5,return_train_score = False)
```

```python
classifire_rdcv.fit(x,y)
```

```python
results_rdcv = pd.DataFrame(classifire_rdcv.cv_results_)
results_rdcv
```
